```python
# ============================================================
# AGE CALCULATION
# ============================================================

def calculate_age(birth_dates, reference_date):
    """
    Calculate completed age as of reference_date.
    """
    return (
        reference_date.year
        - birth_dates.dt.year
        - (
            (reference_date.month, reference_date.day)
            < (birth_dates.dt.month, birth_dates.dt.day)
        )
    )
```

```python
# ============================================================
# AGE CALCULATION - CORRECTED
# ============================================================

def calculate_age(birth_dates, reference_date):

    birth_dates = pd.to_datetime(
        birth_dates,
        errors="coerce"
    )

    reference_date = pd.Timestamp(reference_date)

    age = (
        reference_date.year
        - birth_dates.dt.year
    )

    # Has the employee's birthday occurred yet in this year?
    birthday_not_reached = (
        (birth_dates.dt.month > reference_date.month)
        |
        (
            (birth_dates.dt.month == reference_date.month)
            &
            (birth_dates.dt.day > reference_date.day)
        )
    )

    age = age - birthday_not_reached.astype(int)

    return age
```

```python
# ============================================================
# PREPARE DATA FOR MONTHLY DIVERSITY ANALYSIS
# ============================================================

active_df = employee_master_clean.copy()

# Make sure required columns are datetime
date_cols = [
    "Actual Join Date",
    "Resignation Date",
    "Last Working Date",
    "Birth Date"
]

for col in date_cols:
    active_df[col] = pd.to_datetime(
        active_df[col],
        errors="coerce"
    )

print("Active Data Shape:", active_df.shape)
print(
    "Unique Employees:",
    active_df["Employee Code"].nunique()
)

print("\nRequired columns missing:")
for col in [
    "Employee Code",
    "Gender",
    "Religion",
    "Birth Date",
    "Actual Join Date",
    "Resignation Date",
    "Last Working Date"
]:
    print(
        col,
        "->",
        active_df[col].isna().sum(),
        "missing"
    )
```

```python
# ============================================================
# CREATE MONTHLY DIVERSITY HISTORY
# ============================================================

monthly_diversity = []

for month_end in monthly_dates:

    # ========================================================
    # 1. EMPLOYEE MUST HAVE JOINED BY THIS MONTH
    # ========================================================

    joined = (
        active_df["Actual Join Date"].notna()
        &
        (
            active_df["Actual Join Date"]
            <= month_end
        )
    )

    # ========================================================
    # 2. EMPLOYEE SHOULD NOT HAVE RESIGNED BEFORE MONTH END
    # ========================================================

    resigned = (
        active_df["Resignation Date"].notna()
        &
        (
            active_df["Resignation Date"]
            <= month_end
        )
    )

    # ========================================================
    # 3. EMPLOYEE SHOULD NOT HAVE LAST WORKING DATE
    #    BEFORE / ON MONTH END
    # ========================================================

    left_company = (
        active_df["Last Working Date"].notna()
        &
        (
            active_df["Last Working Date"]
            <= month_end
        )
    )

    # ========================================================
    # ACTIVE EMPLOYEES AS OF MONTH END
    # ========================================================

    active_mask = (
        joined
        &
        ~resigned
        &
        ~left_company
    )

    month_df = active_df.loc[
        active_mask
    ].copy()

    # ========================================================
    # CALCULATE AGE AS OF MONTH END
    # ========================================================

    month_df["Age"] = calculate_age(
        month_df["Birth Date"],
        month_end
    )

    # ========================================================
    # GENDER
    # ========================================================

    gender = (
        month_df["Gender"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    male_count = (
        gender == "male"
    ).sum()

    female_count = (
        gender == "female"
    ).sum()

    other_gender_count = (
        ~gender.isin([
            "male",
            "female"
        ])
        & gender.notna()
    ).sum()

    # ========================================================
    # RELIGION
    # ========================================================

    religion = (
        month_df["Religion"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    hindu_count = (
        religion == "hinduism"
    ).sum()

    muslim_count = (
        religion.isin([
            "islam",
            "muslim"
        ])
    ).sum()

    christian_count = (
        religion.isin([
            "christianity",
            "christian"
        ])
    ).sum()

    other_religion_count = (
        religion.notna()
        &
        ~religion.isin([
            "hinduism",
            "islam",
            "muslim",
            "christianity",
            "christian"
        ])
    ).sum()

    # ========================================================
    # AGE GROUPS
    # ========================================================

    age_below_25 = (
        month_df["Age"] < 25
    ).sum()

    age_25_35 = (
        (month_df["Age"] >= 25)
        &
        (month_df["Age"] < 35)
    ).sum()

    age_35_50 = (
        (month_df["Age"] >= 35)
        &
        (month_df["Age"] <= 50)
    ).sum()

    age_above_50 = (
        month_df["Age"] > 50
    ).sum()

    # ========================================================
    # SAVE MONTH
    # ========================================================

    monthly_diversity.append({

        "ds": month_end,

        "Total Employees": len(month_df),

        "Male": male_count,
        "Female": female_count,
        "Other Gender": other_gender_count,

        "Hinduism": hindu_count,
        "Muslim": muslim_count,
        "Christian": christian_count,
        "Other Religion": other_religion_count,

        "Age Below 25": age_below_25,
        "Age 25-35": age_25_35,
        "Age 35-50": age_35_50,
        "Age Above 50": age_above_50
    })


# ============================================================
# CONVERT TO DATAFRAME
# ============================================================

monthly_diversity = pd.DataFrame(
    monthly_diversity
)

monthly_diversity["ds"] = pd.to_datetime(
    monthly_diversity["ds"]
)

print("Monthly Diversity Shape:")
print(monthly_diversity.shape)

display(monthly_diversity.head())
```

```python
# ============================================================
# MODEL EVALUATION FUNCTION
# ============================================================

import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error


def calculate_metrics(actual, predicted):
    
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    # Avoid division by zero
    non_zero_mask = actual != 0

    if non_zero_mask.sum() > 0:
        mape = np.mean(
            np.abs(
                (
                    actual[non_zero_mask]
                    -
                    predicted[non_zero_mask]
                )
                /
                actual[non_zero_mask]
            )
        ) * 100
    else:
        mape = np.nan

    return mae, rmse, mape
```

```python
# ============================================================
# TRAIN PROPHET MODELS FOR DIVERSITY FACTORS
# ============================================================

from prophet import Prophet

models = {}
test_predictions = {}
model_metrics = []

for column in forecast_columns:

    print("\n" + "=" * 70)
    print("Training model:", column)
    print("=" * 70)

    # --------------------------------------------------------
    # Prepare Prophet training data
    # --------------------------------------------------------

    prophet_train = train_df[
        ["ds", column]
    ].copy()

    prophet_train = prophet_train.rename(
        columns={
            column: "y"
        }
    )

    # Prophet requires numeric target
    prophet_train["y"] = pd.to_numeric(
        prophet_train["y"],
        errors="coerce"
    )

    prophet_train = prophet_train.dropna()

    # --------------------------------------------------------
    # Create model
    # --------------------------------------------------------

    model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        seasonality_mode="additive"
    )

    # --------------------------------------------------------
    # Train
    # --------------------------------------------------------

    model.fit(prophet_train)

    models[column] = model

    # --------------------------------------------------------
    # Predict test period
    # --------------------------------------------------------

    future_test = test_df[
        ["ds"]
    ].copy()

    forecast_test = model.predict(
        future_test
    )

    predictions = forecast_test[
        "yhat"
    ].values

    # Diversity counts cannot be negative
    predictions = np.maximum(
        predictions,
        0
    )

    test_predictions[column] = predictions

    # --------------------------------------------------------
    # Actual values
    # --------------------------------------------------------

    actual = test_df[
        column
    ].values

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    mae, rmse, mape = calculate_metrics(
        actual,
        predictions
    )

    model_metrics.append({

        "Factor": column,

        "MAE": mae,

        "RMSE": rmse,

        "MAPE (%)": mape
    })

    print(
        f"MAE  : {mae:.2f}"
    )

    print(
        f"RMSE : {rmse:.2f}"
    )

    print(
        f"MAPE : {mape:.2f}%"
    )
```

```python
# ============================================================
# MODEL EVALUATION SUMMARY
# ============================================================

model_metrics_df = pd.DataFrame(
    model_metrics
)

model_metrics_df = model_metrics_df.sort_values(
    "MAPE (%)"
)

print("============================================================")
print("DIVERSITY MODEL EVALUATION")
print("============================================================")

display(
    model_metrics_df
)
```

```python
# ============================================================
# FINAL 6-MONTH FORECAST
# ============================================================

FUTURE_MONTHS = 6

future_forecast = pd.DataFrame()

# Create future monthly dates
last_historical_date = (
    diversity_forecast_df["ds"].max()
)

future_dates = pd.date_range(
    start=last_historical_date + pd.offsets.MonthEnd(1),
    periods=FUTURE_MONTHS,
    freq="ME"
)

future_forecast["ds"] = future_dates

print("============================================================")
print("FORECAST PERIOD")
print("============================================================")

print(
    "Historical last month:",
    last_historical_date
)

print(
    "Forecast start:",
    future_dates.min()
)

print(
    "Forecast end:",
    future_dates.max()
)
```

```python
# ============================================================
# TRAIN FINAL MODELS ON COMPLETE HISTORY
# ============================================================

final_models = {}

for column in forecast_columns:

    print(
        f"\nForecasting: {column}"
    )

    # --------------------------------------------------------
    # Use complete historical dataset
    # --------------------------------------------------------

    prophet_full = diversity_forecast_df[
        ["ds", column]
    ].copy()

    prophet_full = prophet_full.rename(
        columns={
            column: "y"
        }
    )

    prophet_full["y"] = pd.to_numeric(
        prophet_full["y"],
        errors="coerce"
    )

    prophet_full = prophet_full.dropna()

    # --------------------------------------------------------
    # Model
    # --------------------------------------------------------

    model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        seasonality_mode="additive"
    )

    model.fit(
        prophet_full
    )

    final_models[column] = model

    # --------------------------------------------------------
    # Future dates
    # --------------------------------------------------------

    future = pd.DataFrame({
        "ds": future_dates
    })

    forecast = model.predict(
        future
    )

    # --------------------------------------------------------
    # Prevent negative employee counts
    # --------------------------------------------------------

    predictions = np.maximum(
        forecast["yhat"].values,
        0
    )

    # Round employee counts
    predictions = np.round(
        predictions
    ).astype(int)

    future_forecast[column] = predictions


print("\n============================================================")
print("FINAL FORECAST GENERATED")
print("============================================================")

display(
    future_forecast
)
```
