```python
# ============================================================
# AGE CALCULATION
# ============================================================

def calculate_age(birth_dates, reference_date):
    """
    Calculate completed age as of reference_date.
    """
    return (
        reference_date.year
        - birth_dates.dt.year
        - (
            (reference_date.month, reference_date.day)
            < (birth_dates.dt.month, birth_dates.dt.day)
        )
    )
```

```python
# ============================================================
# AGE CALCULATION - CORRECTED
# ============================================================

def calculate_age(birth_dates, reference_date):

    birth_dates = pd.to_datetime(
        birth_dates,
        errors="coerce"
    )

    reference_date = pd.Timestamp(reference_date)

    age = (
        reference_date.year
        - birth_dates.dt.year
    )

    # Has the employee's birthday occurred yet in this year?
    birthday_not_reached = (
        (birth_dates.dt.month > reference_date.month)
        |
        (
            (birth_dates.dt.month == reference_date.month)
            &
            (birth_dates.dt.day > reference_date.day)
        )
    )

    age = age - birthday_not_reached.astype(int)

    return age
```

```python
# ============================================================
# PREPARE DATA FOR MONTHLY DIVERSITY ANALYSIS
# ============================================================

active_df = employee_master_clean.copy()

# Make sure required columns are datetime
date_cols = [
    "Actual Join Date",
    "Resignation Date",
    "Last Working Date",
    "Birth Date"
]

for col in date_cols:
    active_df[col] = pd.to_datetime(
        active_df[col],
        errors="coerce"
    )

print("Active Data Shape:", active_df.shape)
print(
    "Unique Employees:",
    active_df["Employee Code"].nunique()
)

print("\nRequired columns missing:")
for col in [
    "Employee Code",
    "Gender",
    "Religion",
    "Birth Date",
    "Actual Join Date",
    "Resignation Date",
    "Last Working Date"
]:
    print(
        col,
        "->",
        active_df[col].isna().sum(),
        "missing"
    )
```

```python
# ============================================================
# CREATE MONTHLY DIVERSITY HISTORY
# ============================================================

monthly_diversity = []

for month_end in monthly_dates:

    # ========================================================
    # 1. EMPLOYEE MUST HAVE JOINED BY THIS MONTH
    # ========================================================

    joined = (
        active_df["Actual Join Date"].notna()
        &
        (
            active_df["Actual Join Date"]
            <= month_end
        )
    )

    # ========================================================
    # 2. EMPLOYEE SHOULD NOT HAVE RESIGNED BEFORE MONTH END
    # ========================================================

    resigned = (
        active_df["Resignation Date"].notna()
        &
        (
            active_df["Resignation Date"]
            <= month_end
        )
    )

    # ========================================================
    # 3. EMPLOYEE SHOULD NOT HAVE LAST WORKING DATE
    #    BEFORE / ON MONTH END
    # ========================================================

    left_company = (
        active_df["Last Working Date"].notna()
        &
        (
            active_df["Last Working Date"]
            <= month_end
        )
    )

    # ========================================================
    # ACTIVE EMPLOYEES AS OF MONTH END
    # ========================================================

    active_mask = (
        joined
        &
        ~resigned
        &
        ~left_company
    )

    month_df = active_df.loc[
        active_mask
    ].copy()

    # ========================================================
    # CALCULATE AGE AS OF MONTH END
    # ========================================================

    month_df["Age"] = calculate_age(
        month_df["Birth Date"],
        month_end
    )

    # ========================================================
    # GENDER
    # ========================================================

    gender = (
        month_df["Gender"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    male_count = (
        gender == "male"
    ).sum()

    female_count = (
        gender == "female"
    ).sum()

    other_gender_count = (
        ~gender.isin([
            "male",
            "female"
        ])
        & gender.notna()
    ).sum()

    # ========================================================
    # RELIGION
    # ========================================================

    religion = (
        month_df["Religion"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    hindu_count = (
        religion == "hinduism"
    ).sum()

    muslim_count = (
        religion.isin([
            "islam",
            "muslim"
        ])
    ).sum()

    christian_count = (
        religion.isin([
            "christianity",
            "christian"
        ])
    ).sum()

    other_religion_count = (
        religion.notna()
        &
        ~religion.isin([
            "hinduism",
            "islam",
            "muslim",
            "christianity",
            "christian"
        ])
    ).sum()

    # ========================================================
    # AGE GROUPS
    # ========================================================

    age_below_25 = (
        month_df["Age"] < 25
    ).sum()

    age_25_35 = (
        (month_df["Age"] >= 25)
        &
        (month_df["Age"] < 35)
    ).sum()

    age_35_50 = (
        (month_df["Age"] >= 35)
        &
        (month_df["Age"] <= 50)
    ).sum()

    age_above_50 = (
        month_df["Age"] > 50
    ).sum()

    # ========================================================
    # SAVE MONTH
    # ========================================================

    monthly_diversity.append({

        "ds": month_end,

        "Total Employees": len(month_df),

        "Male": male_count,
        "Female": female_count,
        "Other Gender": other_gender_count,

        "Hinduism": hindu_count,
        "Muslim": muslim_count,
        "Christian": christian_count,
        "Other Religion": other_religion_count,

        "Age Below 25": age_below_25,
        "Age 25-35": age_25_35,
        "Age 35-50": age_35_50,
        "Age Above 50": age_above_50
    })


# ============================================================
# CONVERT TO DATAFRAME
# ============================================================

monthly_diversity = pd.DataFrame(
    monthly_diversity
)

monthly_diversity["ds"] = pd.to_datetime(
    monthly_diversity["ds"]
)

print("Monthly Diversity Shape:")
print(monthly_diversity.shape)

display(monthly_diversity.head())
```

```python
# ============================================================
# MODEL EVALUATION FUNCTION
# ============================================================

import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error


def calculate_metrics(actual, predicted):
    
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    # Avoid division by zero
    non_zero_mask = actual != 0

    if non_zero_mask.sum() > 0:
        mape = np.mean(
            np.abs(
                (
                    actual[non_zero_mask]
                    -
                    predicted[non_zero_mask]
                )
                /
                actual[non_zero_mask]
            )
        ) * 100
    else:
        mape = np.nan

    return mae, rmse, mape
```

## Forecasting architecture update

The final model forecasts **Total Employees** first using a tuned Prophet configuration selected by chronological holdout performance. Gender, age and religion are then forecast as workforce shares and converted to integer counts using the same total workforce forecast. This prevents demographic groups from producing totals that disagree with the workforce total.

Categories with no historical observations/variation are handled explicitly rather than being reported as if a zero-error Prophet model were evidence of accuracy. Religion missingness is retained as **Religion Not Available** so the religion groups reconcile to the workforce total.

**Important:** MAE, RMSE and MAPE are backtest error metrics; they are not percentage “accuracy” scores. A continuity check flags a large first-month change instead of silently modifying the forecast.


```python
# ============================================================
# TOTAL EMPLOYEES MODEL SELECTION - TUNED PROPHET
# ============================================================
# The previous version used one Prophet configuration and could
# produce an implausibly large drop immediately after Aug-2026.
# Here we tune BOTH recent-history window and changepoint strength
# using the same chronological 12-month holdout. No future data is
# used for model selection.

from prophet import Prophet
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

HISTORY_WINDOWS = [12, 18, 24, 36, 60, 84, 120, len(train_df)]
HISTORY_WINDOWS = sorted(set(
    int(w) for w in HISTORY_WINDOWS
    if int(w) >= 12 and int(w) <= len(train_df)
))

CHANGEPOINT_PRIOR_SCALES = [0.01, 0.05, 0.10, 0.20, 0.50]


def calculate_metrics(actual, predicted):
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))

    non_zero_mask = actual != 0
    if non_zero_mask.sum() > 0:
        mape = (
            np.mean(
                np.abs(
                    (actual[non_zero_mask] - predicted[non_zero_mask])
                    / actual[non_zero_mask]
                )
            ) * 100
        )
    else:
        mape = np.nan

    return mae, rmse, mape


def fit_prophet(train_data, target_col, changepoint_prior_scale=0.05):
    prophet_df = (
        train_data[["ds", target_col]]
        .rename(columns={target_col: "y"})
        .copy()
    )

    prophet_df["y"] = pd.to_numeric(
        prophet_df["y"], errors="coerce"
    )
    prophet_df = prophet_df.dropna()

    model = Prophet(
        growth="linear",
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        seasonality_mode="additive",
        changepoint_prior_scale=changepoint_prior_scale
    )

    model.fit(prophet_df)
    return model


# ------------------------------------------------------------
# GRID SEARCH ON THE CHRONOLOGICAL HOLDOUT
# ------------------------------------------------------------
total_model_results = []

for window in HISTORY_WINDOWS:
    window_train = train_df.tail(window).copy()

    for cps in CHANGEPOINT_PRIOR_SCALES:
        model = fit_prophet(
            window_train,
            "Total Employees",
            changepoint_prior_scale=cps
        )

        prediction = model.predict(test_df[["ds"]])["yhat"].to_numpy()
        prediction = np.maximum(prediction, 0)

        mae, rmse, mape = calculate_metrics(
            test_df["Total Employees"].to_numpy(),
            prediction
        )

        total_model_results.append({
            "History Months": window,
            "Changepoint Prior Scale": cps,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE (%)": mape
        })


total_model_selection = (
    pd.DataFrame(total_model_results)
    .sort_values(["MAPE (%)", "RMSE", "MAE"])
    .reset_index(drop=True)
)

BEST_TOTAL_WINDOW = int(
    total_model_selection.iloc[0]["History Months"]
)
BEST_TOTAL_CPS = float(
    total_model_selection.iloc[0]["Changepoint Prior Scale"]
)

# Keep the old variable name for downstream cells.
BEST_HISTORY_MONTHS = BEST_TOTAL_WINDOW

print("============================================================")
print("TUNED TOTAL EMPLOYEES MODEL SELECTION")
print("============================================================")
display(total_model_selection)

print(
    f"Selected history window: {BEST_TOTAL_WINDOW} months"
)
print(
    f"Selected changepoint prior scale: {BEST_TOTAL_CPS}"
)
```

```python
# ============================================================
# DIVERSITY SHARE MODEL EVALUATION
# ============================================================
# Gender, age and religion are forecast as proportions of the
# total workforce. Counts are reconstructed only after the total
# workforce forecast is generated. This guarantees reconciliation.

gender_columns = [
    "Male",
    "Female",
    "Other Gender"
]

age_columns = [
    "Age Below 25",
    "Age 25-35",
    "Age 35-50",
    "Age Above 50"
]

religion_columns = [
    "Hinduism",
    "Muslim",
    "Christian",
    "Other Religion",
    "Religion Not Available"
]

share_groups = {
    "Gender": gender_columns,
    "Age": age_columns,
    "Religion": religion_columns
}


def prepare_share_data(data, column):
    tmp = data[["ds", column, "Total Employees"]].copy()
    tmp["share"] = np.where(
        tmp["Total Employees"] > 0,
        tmp[column] / tmp["Total Employees"],
        0
    )
    return tmp[["ds", "share"]].rename(columns={"share": "y"})


def fit_share_model(
    data,
    column,
    changepoint_prior_scale=0.05
):
    share_df = prepare_share_data(data, column)

    # If a category has no historical observations, do not fit a
    # meaningless Prophet model. Its forecast is explicitly zero.
    if np.isclose(share_df["y"].abs().sum(), 0):
        return None

    model = Prophet(
        growth="linear",
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        seasonality_mode="additive",
        changepoint_prior_scale=changepoint_prior_scale
    )
    model.fit(share_df)
    return model


share_metrics = []
share_train = train_df.tail(BEST_HISTORY_MONTHS).copy()

for group_name, columns in share_groups.items():
    for column in columns:
        model = fit_share_model(
            share_train,
            column,
            changepoint_prior_scale=BEST_TOTAL_CPS
        )

        if model is None:
            predicted_share = np.zeros(len(test_df))
            model_type = "No historical variation - zero"
        else:
            predicted_share = model.predict(test_df[["ds"]])["yhat"].to_numpy()
            predicted_share = np.clip(predicted_share, 0, 1)
            model_type = "Prophet - demographic share"

        predicted_count = (
            predicted_share * test_df["Total Employees"].to_numpy()
        )
        actual_count = test_df[column].to_numpy()

        mae, rmse, mape = calculate_metrics(
            actual_count, predicted_count
        )

        share_metrics.append({
            "Group": group_name,
            "Factor": column,
            "Model Type": model_type,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE (%)": mape
        })

share_metrics_df = pd.DataFrame(share_metrics)

print("============================================================")
print("DIVERSITY SHARE MODEL EVALUATION")
print("============================================================")
display(share_metrics_df)
```

```python
# ============================================================
# FINAL 6-MONTH FORECAST PERIOD
# ============================================================

FUTURE_MONTHS = 6

last_historical_date = (
    diversity_forecast_df["ds"].max()
)

future_dates = pd.date_range(
    start=last_historical_date
        + pd.offsets.MonthEnd(1),
    periods=FUTURE_MONTHS,
    freq="ME"
)

print("============================================================")
print("FORECAST PERIOD")
print("============================================================")

print(
    "Historical last month:",
    last_historical_date
)

print(
    "Forecast start:",
    future_dates.min()
)

print(
    "Forecast end:",
    future_dates.max()
)
```

```python
# ============================================================
# FINAL FORECAST
# ============================================================
# 1. Forecast Total Employees using the selected Prophet setup.
# 2. Forecast demographic shares.
# 3. Normalize shares within each demographic group.
# 4. Allocate integer employees so every group reconciles exactly
#    to Total Employees.

final_train = diversity_forecast_df.tail(BEST_TOTAL_WINDOW).copy()


def allocate_integer_counts(total, raw_values):
    """Allocate integer counts using the largest-remainder method."""
    total = int(max(0, round(total)))
    values = np.asarray(raw_values, dtype=float)
    values = np.clip(values, 0, None)
    value_sum = values.sum()

    if value_sum <= 0:
        return np.zeros(len(values), dtype=int)

    shares = values / value_sum
    exact = total * shares
    base = np.floor(exact).astype(int)
    remaining = total - base.sum()

    if remaining > 0:
        order = np.argsort(-(exact - base))
        base[order[:remaining]] += 1

    return base


# ------------------------------------------------------------
# TOTAL EMPLOYEES
# ------------------------------------------------------------
total_model = fit_prophet(
    final_train,
    "Total Employees",
    changepoint_prior_scale=BEST_TOTAL_CPS
)

total_prediction = total_model.predict(
    pd.DataFrame({"ds": future_dates})
)["yhat"].to_numpy()

total_prediction = np.maximum(total_prediction, 0)
total_prediction = np.round(total_prediction).astype(int)

future_forecast = pd.DataFrame({
    "ds": future_dates,
    "Total Employees": total_prediction
})


# ------------------------------------------------------------
# DEMOGRAPHIC GROUPS
# ------------------------------------------------------------
for group_name, columns in share_groups.items():
    predicted_shares = []

    for column in columns:
        model = fit_share_model(
            final_train,
            column,
            changepoint_prior_scale=BEST_TOTAL_CPS
        )

        if model is None:
            share = np.zeros(len(future_dates))
        else:
            share = model.predict(
                pd.DataFrame({"ds": future_dates})
            )["yhat"].to_numpy()
            share = np.clip(share, 0, 1)

        predicted_shares.append(share)

    predicted_shares = np.column_stack(predicted_shares)

    row_sums = predicted_shares.sum(axis=1, keepdims=True)
    normalized_shares = np.divide(
        predicted_shares,
        row_sums,
        out=np.zeros_like(predicted_shares),
        where=row_sums > 0
    )

    for i in range(len(future_dates)):
        counts = allocate_integer_counts(
            future_forecast.loc[i, "Total Employees"],
            normalized_shares[i]
        )

        for j, column in enumerate(columns):
            future_forecast.loc[i, column] = counts[j]


for column in gender_columns + age_columns + religion_columns:
    future_forecast[column] = (
        future_forecast[column].fillna(0).astype(int)
    )

print("============================================================")
print("FINAL FORECAST GENERATED")
print("============================================================")
display(future_forecast)
```

```python
# ============================================================
# FINAL CONSISTENCY AND FORECAST SANITY CHECKS
# ============================================================

future_forecast["Gender Total"] = (
    future_forecast["Male"]
    + future_forecast["Female"]
    + future_forecast["Other Gender"]
)
future_forecast["Gender Difference"] = (
    future_forecast["Total Employees"]
    - future_forecast["Gender Total"]
)

future_forecast["Age Total"] = (
    future_forecast["Age Below 25"]
    + future_forecast["Age 25-35"]
    + future_forecast["Age 35-50"]
    + future_forecast["Age Above 50"]
)
future_forecast["Age Difference"] = (
    future_forecast["Total Employees"]
    - future_forecast["Age Total"]
)

future_forecast["Religion Total"] = (
    future_forecast["Hinduism"]
    + future_forecast["Muslim"]
    + future_forecast["Christian"]
    + future_forecast["Other Religion"]
    + future_forecast["Religion Not Available"]
)
future_forecast["Religion Difference"] = (
    future_forecast["Total Employees"]
    - future_forecast["Religion Total"]
)

last_actual_total = int(
    diversity_forecast_df.iloc[-1]["Total Employees"]
)
first_forecast_total = int(
    future_forecast.iloc[0]["Total Employees"]
)

first_month_change_pct = (
    (first_forecast_total - last_actual_total)
    / last_actual_total * 100
    if last_actual_total else np.nan
)

future_forecast["Change vs Previous Actual (%)"] = np.nan
future_forecast.loc[0, "Change vs Previous Actual (%)"] = first_month_change_pct

print("============================================================")
print("FINAL CONSISTENCY CHECK")
print("============================================================")

print(
    "Gender consistency:",
    (future_forecast["Gender Difference"] == 0).all()
)
print(
    "Age consistency:",
    (future_forecast["Age Difference"] == 0).all()
)
print(
    "Religion consistency:",
    (future_forecast["Religion Difference"] == 0).all()
)
print(
    f"Latest actual Total Employees: {last_actual_total:,}"
)
print(
    f"First forecast Total Employees: {first_forecast_total:,}"
)
print(
    f"First-month change vs latest actual: {first_month_change_pct:.2f}%"
)

if abs(first_month_change_pct) > 15:
    print(
        "WARNING: first-month workforce change exceeds 15%. "
        "Review the selected Prophet model and source trend before business use."
    )
else:
    print("Forecast continuity check: PASS")

display(
    future_forecast[
        ["ds", "Total Employees", "Male", "Female", "Other Gender",
         "Gender Total", "Gender Difference"]
    ]
)

display(
    future_forecast[
        ["ds", "Total Employees", "Age Below 25", "Age 25-35",
         "Age 35-50", "Age Above 50", "Age Total", "Age Difference"]
    ]
)

display(
    future_forecast[
        ["ds", "Total Employees", "Hinduism", "Muslim", "Christian",
         "Other Religion", "Religion Not Available", "Religion Total",
         "Religion Difference"]
    ]
)
```
